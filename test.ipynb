```python
from signal_sythesis import generate_transition_signal
import matplotlib.pyplot as plt

t, pressure, wave_count = generate_transition_signal(
    duration_s=0.02,
    sample_rate_hz=2_000_000,
    rotation_freq_hz=10000,   # changed from 5000 to 10000
    transition_schedule=[(0.0, 1), (0.005, 2), (0.012, 1)],
    noise_std=0.02,
    seed=42,
)

fig, axes = plt.subplots(2, 1, figsize=(10, 6), sharex=True)
axes[0].plot(t * 1e3, pressure, linewidth=0.5)
axes[0].set_ylabel("Pressure (a.u.)")
axes[0].set_title("10 kHz Rotation Frequency")
axes[1].plot(t * 1e3, wave_count, drawstyle="steps-post", color="orange")
axes[1].set_ylabel("Wave count")
axes[1].set_xlabel("Time (ms)")
plt.tight_layout()
plt.show()
```

```python
from signal_sythesis import generate_transition_signal
import matplotlib.pyplot as plt

t, pressure, wave_count = generate_transition_signal(
    duration_s=0.02,
    sample_rate_hz=2_000_000,
    rotation_freq_hz=5000,
    transition_schedule=[(0.0, 1), (0.005, 2), (0.010, 3), (0.015, 1)],  # added 3-wave segment
    noise_std=0.02,
    seed=42,
)

fig, axes = plt.subplots(2, 1, figsize=(10, 6), sharex=True)
axes[0].plot(t * 1e3, pressure, linewidth=0.5)
axes[0].set_ylabel("Pressure (a.u.)")
axes[0].set_title("1 -> 2 -> 3 -> 1 Wave Transitions")
axes[1].plot(t * 1e3, wave_count, drawstyle="steps-post", color="orange")
axes[1].set_ylabel("Wave count")
axes[1].set_xlabel("Time (ms)")
axes[1].set_yticks([0, 1, 2, 3])
plt.tight_layout()
plt.show()
```

```python
from signal_sythesis import generate_transition_signal
import matplotlib.pyplot as plt

t, pressure, wave_count = generate_transition_signal(
    duration_s=0.02,
    sample_rate_hz=2_000_000,
    rotation_freq_hz=5000,
    transition_schedule=[(0.0, 1), (0.005, 2), (0.012, 1)],
    noise_std=0.15,   # increased from 0.02 to 0.15
    seed=42,
)

fig, axes = plt.subplots(2, 1, figsize=(10, 6), sharex=True)
axes[0].plot(t * 1e3, pressure, linewidth=0.5)
axes[0].set_ylabel("Pressure (a.u.)")
axes[0].set_title("High Noise (noise_std=0.15)")
axes[1].plot(t * 1e3, wave_count, drawstyle="steps-post", color="orange")
axes[1].set_ylabel("Wave count")
axes[1].set_xlabel("Time (ms)")
plt.tight_layout()
plt.show()
```
